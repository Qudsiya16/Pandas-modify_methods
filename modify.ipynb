```python
import pandas as pd

data = {
    "Name": ["Ali", "Ayesha", "Amjad", "Sana", "Usman", "Hina", "Ahmed", "Zoya"],
    "Age": [20, 22, 21, 23, 19, 24, 22, 20],
    "Marks": [85, 67, 45, 92, 38, 76, 55, 88],
    "City": ["Karachi", "Lahore", "Karachi", "Multan", "Lahore", "Multan", "Karachi", "Lahore"],
    "Grade": ["A", "B", "C", "A", "D", "B", "C", "A"]
}

a = pd.DataFrame(data)
a
```

```python
#to detect data 1 by 1 and check the conditions
a["Result"] = a["Marks"].apply(lambda x:"Pass" if x > 65 else "Fail")
a
```

```python
a["Status"] = a["Age"].apply(lambda x: "Young" if x< 20 else "Adult")
a
```

```python
#.map() -- update the value
#a["Grade"] = a["Grade"].map({"A": "A+", "B": "B++"}) ---shows missing values to those that are not given to  modify.
a["Grade"] = a["Grade"].map({"A": "A+", "B": "B++","C": "C", "D": "D"})
a
```

```python
#.replace() -- modify the given values and keeps the existing values as it is.
a["City"] = a["City"].replace({"Karachi": "KPK"})
a
```
